# Plumbing part starter

This notebook is a starting point for building plumbing parts with CadQuery.
Install `cadquery` in the notebook environment before running the code cell below.

```python
import cadquery as cq

from plumbing_3d import starter_pipe_dimensions

dimensions = starter_pipe_dimensions()
pipe = (
    cq.Workplane("XY")
    .circle(dimensions.outer_diameter_mm / 2)
    .circle(dimensions.inner_diameter_mm / 2)
    .extrude(dimensions.length_mm)
)

pipe
```
